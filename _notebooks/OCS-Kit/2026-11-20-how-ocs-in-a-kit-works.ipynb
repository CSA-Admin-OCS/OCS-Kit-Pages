## What OCS-in-a-Kit is

<div class="ocskit"><div class="ocskit-text"><p>Open Coding Society runs a live website and two backend servers for its own school. That code holds school data, student information and API keys, so another school can't safely copy it.</p><p>OCS-in-a-Kit is a cleaned copy of the same platform. The school-specific parts are taken out, the features are grouped into kits, and an installer sets the whole thing up on a new school's own server.</p></div><div class="ocskit-grid ocskit-grid--4"><div class="ocskit-card ocskit-card--accent"><b class="ocskit-h">Website</b><p>Jekyll site on GitHub Pages: lessons, guides, profile, admin dashboard. Repo: <code>OCS-Kit-Pages</code>.</p></div><div class="ocskit-card ocskit-card--accent"><b class="ocskit-h">Flask backend</b><p>Python API: logins, sections, AI chat proxy, code runner, social. Repo: <code>OCS-Kit-Flask</code>.</p></div><div class="ocskit-card ocskit-card--accent"><b class="ocskit-h">Spring backend</b><p>Java API: people, groups, assignments, calendar, hall pass. Repo: <code>OCS-Kit-Spring</code>.</p></div><div class="ocskit-card ocskit-card--accent"><b class="ocskit-h">Installer</b><p>One command asks a few questions, writes the settings and starts everything. Repo: <code>OCS-Kit-Vanilla</code>.</p></div></div><div class="ocskit-explain"><span class="ocskit-label">How to explain it</span><ul><li>It's our platform with the school taken out, so any school can run its own copy.</li><li>There are four parts: the website, two backends, and an installer that ties them together.</li><li>A school runs one command and answers a few questions. It never has to read our code.</li></ul></div></div>

## A running kit

<div class="ocskit"><div class="ocskit-text"><p>The website is only static files, served free by GitHub Pages. When someone signs in or opens anything with data, the page's JavaScript calls the backends.</p><p>The backends run in Docker on a server the school controls. nginx sits in front and handles HTTPS. Student code never runs inside a backend: it is sent to separate runner containers that have no internet access and strict memory and CPU limits.</p></div><figure class="ocskit-diagram"><div class="ocskit-canvas"><svg viewBox="0 0 1000 440" role="img" aria-label="The browser loads static pages from GitHub Pages, then calls the backends through nginx on your server. nginx forwards to flask, socket and spring containers. Flask and Spring send student code to runner containers on an internal network with no internet. Images are pulled from Docker Hub.">

## Turning a feature on or off

<div class="ocskit"><div class="ocskit-text"><p>Every feature is listed once, in <code>kits/catalog.json</code>. Each backend works out which features are on: it starts from the kit, then applies any switches an Admin flipped in the dashboard.</p><p>When a feature is off, its backend routes answer 404 and the website hides its links. Hiding links is for convenience. The real protection is the backend refusing the request.</p></div><figure class="ocskit-diagram"><div class="ocskit-canvas"><svg viewBox="0 0 1000 340" role="img" aria-label="catalog.json defines features and kit presets. Each backend starts from the kit an Admin stored in the database, or otherwise from KIT and FEATURES in its .env file, then applies the Admin's per-feature overrides. The result decides which routes answer 404, and the website reads it from /api/kit on both backends to hide links.">

## Where the code comes from

<div class="ocskit"><div class="ocskit-text"><p>New work is built in the live Open Coding Society repos. It moves into the kit through an import tool that copies one feature at a time and runs a scrub check on every file.</p><p>The scrub check blocks school names, keys and tokens, student data files and oversized files. If it finds anything, nothing is copied. What passes lands on a review branch, and a person merges it.</p></div><figure class="ocskit-diagram"><div class="ocskit-canvas"><svg viewBox="0 0 1000 360" role="img" aria-label="Upstream pages, flask and spring repos are scanned weekly by OCS-Kit/sync, which runs a scrub check and opens pull requests into the four OCS-Kit repos. Those produce the installer, the GitHub Pages website and Docker Hub images. CSA-Admin-OCS holds forks of the kit repos.">

## Flask only or Flask + Spring

<div class="ocskit"><div class="ocskit-text"><p>A teammate already solved part of packing out upstream. One <code>spring</code> setting in <code>_config.yml</code> decides whether the site uses Flask alone or Flask and Spring (<a href="https://github.com/Open-Coding-Society/pages/pull/1844">pages #1844</a>). Instead of checking <code>if (springEnabled)</code> all over the code, the site picks a class once: <code>FlaskSession</code>, or <code>SpringSession</code>, which extends it.</p><p>In his review, Mr. Mortensen called the switch the right way to manage shared code and the inheritance technique "far superior". He sees kits as a combination of three things, labelled A, B and C below. The kit will adopt this. Today the kit always assumes both backends and still has the older login and profile.</p></div><div class="ocskit-compare" style="margin-bottom:1rem"><div class="ocskit-option"><header><code>spring: false</code><b class="ocskit-h">Flask only</b></header><p>Single login. Login, navbar and Profile use <code>FlaskSession</code> and <code>FlaskProfile</code> and never call Spring. Spring-only parts like Face ID are left out.</p><div class="ocskit-containers"><span>flask</span><span>socket</span><span>flask_runner</span><span class="off">spring</span><span class="off">spring_runner</span></div><span class="ocskit-label">3 containers · likely Vanilla</span></div><div class="ocskit-option ocskit-option--chosen"><header><code>spring: true</code><b class="ocskit-h">Flask + Spring</b></header><p>Dual login. <code>SpringSession</code> extends <code>FlaskSession</code>: it does every Flask step, then adds Spring's. A Spring failure never undoes the Flask account.</p><div class="ocskit-containers"><span>flask</span><span>socket</span><span>flask_runner</span><span>spring</span><span>spring_runner</span></div><span class="ocskit-label">5 containers · Education, Business today</span></div></div><figure class="ocskit-diagram"><div class="ocskit-canvas"><svg viewBox="0 0 1000 290" role="img" aria-label="The kit choice sets spring true or false in _config.yml. config.js turns it into springEnabled. createBackends returns a Flask backend and a Spring backend, or null for Spring. createSession then picks SpringSession, which extends FlaskSession, or FlaskSession alone, once.
