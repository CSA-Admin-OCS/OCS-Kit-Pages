## Reference Guide

| Term | Definition / Example | College Board Requirement |
|---|---|---|
| **Output** | Information the program shows the user: `System.out.println("Hi");` | Unit 1, Topic 1.3, 1.3.A.1, pp. 33–34 |
| **String literal** | Text written inside double quotes: `"Hello"` | Unit 1, Topic 1.3, pp. 33–34 |
| **Escape sequence** | A backslash plus a character, for characters a string cannot hold directly: `"Line 1\nLine 2"` | Unit 1, Topic 1.3, pp. 33–34. Only `\"`, `\\`, and `\n` are tested. |
| **Arithmetic expression** | Numbers and operators that work out to one value: `2 + 3 * 4` | Unit 1, Topic 1.3, 1.3.C.5, pp. 33–34 |
| **Integer division** | `int / int` keeps only the whole-number part: `7 / 2` is `3` | Unit 1, Topic 1.3, 1.3.C.3, pp. 33–34 |
| **Remainder** | `%` gives what is left over after dividing: `7 % 2` is `1` | Unit 1, Topic 1.3, pp. 33–34 |
| **Concatenation** | `+` joins a String with another value: `"Age: " + 16` | — |

- `+` adds numbers but joins Strings, and Java works left to right
- Dividing an `int` by `0` compiles, then throws an `ArithmeticException`

### Key Takeaways

- Output = communication from program to user.
- `print` vs `println` vs `printf` control formatting.
- Escape sequences handle special characters and formatting.
- Arithmetic expressions make output dynamic.
- `Menu.java` (Example B below) demonstrates real-world application of expressions + output.

### Output Methods

| Method | What it does | Example | Prints |
| --- | --- | --- | --- |
| `print` | Prints and stays on the same line | `System.out.print("Hi ");` | `Hi ` |
| `println` | Prints, then moves to a new line | `System.out.println("Hi");` | `Hi`, then a new line |
| `printf` | Prints using a format pattern (not on the AP exam) | `System.out.printf("%.2f", 3.14159);` | `3.14` |

### Escape Sequences

| Sequence | Meaning | Example | Prints |
| --- | --- | --- | --- |
| `\"` | Double quote | `"She said \"hi\""` | `She said "hi"` |
| `\\` | Backslash | `"C:\\Users"` | `C:\Users` |
| `\n` | New line | `"A\nB"` | `A` and `B` on two lines |
| `\t` | Tab (not on the AP exam) | `"A\tB"` | `A` and `B` with a tab between |

### Operator Precedence

| Order | Operators | Example | Result |
| --- | --- | --- | --- |
| 1st | `( )` | `(2 + 3) * 4` | `20` |
| 2nd | `*`, `/`, `%`, left to right | `2 + 3 * 4` | `14` |
| 3rd | `+`, `-`, left to right | `10 - 4 - 3` | `3` |

### Predicting Output

- **Both values are `int`** → the result is an `int`, decimals dropped
- **At least one `double`** → the result is a `double`
- **A String comes first with `+`** → everything after it is joined as text, unless it is in parentheses
- **`print`** → the next output stays on the same line; **`println`** → the next output starts a new line

## Real-Life Application
### The Stock Index That Lost Half Its Value by Rounding Down

In 1982, the Vancouver Stock Exchange started a new index at 1000. Its computer recalculated the index thousands of times a day and kept only three decimal places, chopping off the rest instead of rounding. Nothing crashed. Nothing looked wrong on any single update. Everyone guess: after about 22 months, what did the index read, and what should it have read?

| Real Case | What Happened | Java Idea in This Lesson |
|:---:|:---|:---|
| Vancouver Stock Exchange (1982 to 1983) | Each update threw away a tiny bit. By late 1983 the index read about 525, when the correct value was about 1099. | Integer division also throws away the decimal part: `7 / 2` is `3`, every time |
| Store receipts | `$4.5` looks broken; `$4.50` looks right | `printf("%.2f", 4.5)` prints exactly two decimal places |
| Windows file paths | `C:\Users\Student` | A backslash inside a String must be written `\\` |
| Text messages with quotes | `She said "hi"` | A quote inside a String must be written `\"` |

**The takeaway:** predict the exact output before you run, especially when division or rounding is involved. Small errors repeated thousands of times add up.

## Tech Talk

**Learning Objective:** Generate output, use string literals and escape sequences, and work out the result of arithmetic expressions.

**Success Criteria:** You can predict what print and println show, fix strings that need escape sequences, and evaluate expressions that use integer division, %, precedence, and string joining.

`print` stays on the same line. `println` moves to the next line.

Two int values always give an int, so division throws away the decimal part. One double changes that.

<pre><code class="language-java">
System.out.println(7 / 3);       // 2
System.out.println(7.0 / 3);     // 2.3333333333333335
System.out.println(7 % 3);       // 1  (remainder)
System.out.println(2 + 3 * 4);   // 14 (* before +)
</code></pre>

`+` joins strings, and Java works left to right:

<pre><code class="language-java">
System.out.println("Result: " + 5 + 3);     // Result: 53
System.out.println("Result: " + (5 + 3));   // Result: 8
</code></pre>

Dividing an int by 0 does not fail at compile time. It throws an `ArithmeticException` while running.

### Code Examples

#### A. print vs println. The PrintDemo class shows which output stays on the line and which moves down

The two get mixed up a lot:

<pre><code class="language-java">
System.out.print("Hello");     // print: the next output stays on this line
System.out.println("Hello");   // println: the next output starts on a new line
</code></pre>

Terminology:

**Output** is text the program sends to the console. `print` and `println` send the same text; the only difference is whether the cursor moves to a new line afterward.

---

```java
// CODE_RUNNER: Predict how many lines this prints, then run it
public class PrintDemo {
    public static void main(String[] args) {
        // Example: println vs print
        System.out.print("Hello");
        System.out.print(" World");
        System.out.println("!"); // moves to new line
        System.out.println("Done");
    }
}

PrintDemo.main(null);
```

#### B. Menu.java. The MenuDemo class combines literals, expressions, and output in one menu

A menu uses literals, expressions, and output together. The runner has no keyboard, so this copy reads its choice from a String. Change `"2\n"` to another option and run it again.

Common joining mistake:

<pre><code class="language-java">
"There are " + 3 + 1 + " options"     // String first, so + joins: There are 31 options
"There are " + (3 + 1) + " options"   // parentheses add first: There are 4 options
</code></pre>

Terminology:

**Concatenation:** once a String is on the left, `+` joins instead of adding. Parentheses make the math happen first.

---

```java
// CODE_RUNNER: Change the simulated choice, then run it
import java.util.Scanner;

public class MenuDemo {
    public static void main(String[] args) {
        Scanner sc = new Scanner("2\n");  // simulated keyboard input

        System.out.println("==== Main Menu ====");
        System.out.println("1. Start Game");
        System.out.println("2. Instructions");
        System.out.println("3. Exit");

        System.out.print("Choose an option: ");
        int choice = sc.nextInt();

        System.out.println("You selected option: " + choice);

        int optionCount = 3;
        System.out.println("There are " + optionCount + " total options.");
    }
}

MenuDemo.main(null);
```
