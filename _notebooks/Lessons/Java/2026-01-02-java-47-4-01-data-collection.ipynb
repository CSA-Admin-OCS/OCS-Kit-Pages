<h4>Slide 2: Core Concepts</h4>

<ul>
    <li>
        <strong>Privacy &amp; Data Ownership:</strong>
        Who owns user data once collected? What risks exist when sensitive data is breached or sold?
    </li>
    <li>
        <strong>Algorithmic Bias:</strong>
        Automated systems in hiring, lending, and law enforcement reinforcing historical biases.
    </li>
    <li>
        <strong>Data Set Fitness:</strong>
        Using incomplete, unrepresentative, or skewed datasets leads to incorrect conclusions.
    </li>
    <li>
        <strong>Teacher Script:</strong>
        "When we talk about 'fitness' in computer science, we mean: Is this data suitable for the purpose of this algorithm? If you train a model on data that only represents a subset of the population, your algorithm will fail when applied to everyone else."
    </li>
</ul>

<h4>Slide 3: Interactive Case Study — "The Flawed Fitness App"</h4>

<ul>
