## Reference Guide

### Key Ideas

- **Nested iteration** is a loop inside the body of another loop.
- The inner loop runs **completely** (all of its passes) for **each single pass** of the outer loop.
- When the loops are independent, the inner body runs `outer count * inner count` times. 3 rows of 4 columns is 12 cells.
- The inner loop can depend on the outer variable, for example `j <= i`. That makes triangles instead of rectangles.
- `System.out.print` stays on the same line; `System.out.println()` after the inner loop ends the row.
- Trace nested loops by tracking the outer and inner variables separately.

### Rectangle Versus Triangle

| Rectangle (independent) | Triangle (dependent) |
| --- | --- |
| `for (int i = 1; i <= 3; i++) {`<br>&nbsp;&nbsp;`for (int j = 1; j <= 4; j++) {`<br>&nbsp;&nbsp;&nbsp;&nbsp;`System.out.print("*");`<br>&nbsp;&nbsp;`}`<br>&nbsp;&nbsp;`System.out.println();`<br>`}` | `for (int i = 1; i <= 3; i++) {`<br>&nbsp;&nbsp;`for (int j = 1; j <= i; j++) {`<br>&nbsp;&nbsp;&nbsp;&nbsp;`System.out.print("*");`<br>&nbsp;&nbsp;`}`<br>&nbsp;&nbsp;`System.out.println();`<br>`}` |
| `****`<br>`****`<br>`****` (12 stars) | `*`<br>`**`<br>`***` (6 stars) |

## Lesson Plan

**Learning Objective:** Represent nested iterative processes.

**Success Criteria:** You can put one loop inside another, state exactly how many times the inner body runs, and build a rectangular or triangular pattern from the two loop variables.

### Tech Talk & Introduction

A nested loop is a loop whose body contains another loop. The inner loop finishes completely on every single pass of the outer loop.

- For an outer loop of `m` passes and an inner loop of `n` passes, the inner body runs `m * n` times.
- When the inner bound depends on the outer variable, the total is the sum of the inner counts, not a simple product. That is what produces triangles instead of rectangles.
- `System.out.print` keeps output on one line; `System.out.println()` after the inner loop ends the row.

## Code Examples

### A. Simple: Nested Loops in Code

A multiplication table. Run it, then change `size` to 5 and 10.

```java
// CODE_RUNNER: Multiplication table: the inner loop prints one row, the outer loop repeats it.
public class TimesTable {
    public static void main(String[] args) {
        int size = 6;   // try 5 and 10

        for (int row = 1; row <= size; row++) {
            for (int col = 1; col <= size; col++) {
                int product = row * col;
                if (product < 10) {
                    System.out.print(" ");   // pad single digits so columns line up
                }
                System.out.print(product + " ");
            }
            System.out.println();
        }
    }
}
TimesTable.main(null);
```

U2.live('u2-11-live', {
  inputs: [
    { name: 'rows', label: 'rows', type: 'range', min: 1, max: 8, value: 4 },
    { name: 'shape', label: 'shape', type: 'select', options: ['square', 'triangle', 'reverse triangle'], value: 'square' }
  ],
  code: [
    'for (int i = 1; i <= rows; i++) {',
    '    for (int j = 1; j <= rows; j++) {          // square',
    '    for (int j = 1; j <= i; j++) {             // triangle',
    '    for (int j = 1; j <= rows - i + 1; j++) {  // reverse triangle',
    '        System.out.print("*");',
    '    }',
    '    System.out.println();',
    '}'
  ],
  evaluate: function (v) {
    var lines = [];
    var stars = 0;
    var active = v.shape === 'square' ? 1 : (v.shape === 'triangle' ? 2 : 3);
    var skipped = [1, 2, 3].filter(function (n) { return n !== active; });
    for (var i = 1; i <= v.rows; i++) {
      var count = v.shape === 'square' ? v.rows : (v.shape === 'triangle' ? i : v.rows - i + 1);
      var row = '';
      for (var j = 1; j <= count; j++) { row += '*'; stars++; }
      lines.push(row);
    }
    var why = v.shape === 'square'
      ? 'Every row has <strong>rows</strong> stars, so the total is rows * rows = ' + stars + '.'
      : (v.shape === 'triangle'
        ? 'Row i has <strong>i</strong> stars: 1 + 2 + ... + ' + v.rows + ' = ' + stars + '.'
        : 'Row i has <strong>rows - i + 1</strong> stars, so the rows shrink: total ' + stars + '.');
    return { lines: [0, active, 4, 6], skipped: skipped, output: lines, explain: 'Only one of the three inner loop headers is used at a time. ' + why };
  }
});
</script>
