## Reference Guide

### Key Topics

<br>

| Term | Definition | Example |
|---|---|---|
| **ArrayList** | A list of objects that grows and shrinks. | `ArrayList<String> names = new ArrayList<String>();` |
| **add** | Appends to the end, or inserts at an index. | `names.add("Ana");` `names.add(0, "Bo");` |
| **get** | Reads the element at an index. | `names.get(0)` |
| **set** | Replaces the element at an index, returns the old one. | `names.set(1, "Cy");` |
| **remove** | Deletes by index or by value, shifting later elements left. | `names.remove(0);` |
| **size** | Number of elements right now. | `names.size()` |

- Valid indexes run from `0` to `size() - 1`. Anything else throws `IndexOutOfBoundsException`.
- An `ArrayList` holds objects, so use `ArrayList<Integer>`, never `ArrayList<int>`.

### Array vs. ArrayList

| | Array | ArrayList |
|---|---|---|
| Size | Fixed | Grows and shrinks |
| Count | `arr.length` | `list.size()` |
| Read / write | `arr[i]` / `arr[i] = v` | `list.get(i)` / `list.set(i, v)` |

### Picking a remove

- **Remove by position** → `nums.remove(2)`
- **Remove by value** → `nums.remove(Integer.valueOf(2))`

## Lesson Plan

**Learning Objective:** Build and change an `ArrayList` with `add`, `get`, `set`, `remove`, and `size`.

**Success Criteria:** You can predict a list after each method call, and choose `remove(int index)` or `remove(Object obj)` on purpose.

### Tech Talk

An array is a row of fixed boxes. An `ArrayList` is a grocery list: it grows when you add and shrinks when you cross something off.

```java
ArrayList<String> groceries = new ArrayList<String>();
groceries.add("Milk");            // [Milk]
groceries.add(0, "Eggs");         // [Eggs, Milk]
groceries.set(1, "Oat Milk");     // [Eggs, Oat Milk]
groceries.remove(0);              // [Oat Milk]
groceries.size();                 // 1
```

## Code Examples

### A. List basics.  The ListBasics class adds, reads, replaces, and counts

Array habits that break:

<pre><code class="language-java">
scores[1] = 95;        // does not compile, use scores.set(1, 95)
scores.length          // does not compile, use scores.size()
</code></pre>

Terminology:

`get` and `set` never change the size. Only `add` and `remove` do.

---

````java
// CODE_RUNNER: Predict each line before running, then change get(3) to get(2)
// add, get, set, size
import java.util.ArrayList;

public class ListBasics {
    public static void main(String[] args) {
        ArrayList<Integer> scores = new ArrayList<Integer>();
        scores.add(70);
        scores.add(90);
        scores.add(1, 85);                 // insert, 90 shifts right

        System.out.println(scores + ", size " + scores.size());
        int old = scores.set(1, 95);       // returns the replaced value
        System.out.println("replaced " + old + " -> " + scores);

        try {
            System.out.println(scores.get(3));
        } catch (IndexOutOfBoundsException e) {
            System.out.println("get(3) failed: " + e.getMessage());
        }
    }
}

ListBasics.main(null);
````

### B. Two remove methods.  The RemoveComparison class removes by position and by value

Common remove mistake:

<pre><code class="language-java">
nums.remove(1);                    // int: removes the element AT index 1
nums.remove(Integer.valueOf(1));   // Integer: removes the VALUE 1
</code></pre>

Terminology:

Java picks the method from the argument type. An `int` means index; an `Integer` means value.

---

````java
// CODE_RUNNER: Predict both lists before running, then try removing Integer.valueOf(9)
// remove(int index) vs. remove(Object obj)
import java.util.ArrayList;

public class RemoveComparison {
    public static void main(String[] args) {
        ArrayList<Integer> byIndex = new ArrayList<Integer>();
        byIndex.add(1);
        byIndex.add(2);
        byIndex.add(3);
        byIndex.remove(1);
        System.out.println("byIndex = " + byIndex);

        ArrayList<Integer> byValue = new ArrayList<Integer>();
        byValue.add(1);
        byValue.add(2);
        byValue.add(3);
        byValue.remove(Integer.valueOf(1));
        System.out.println("byValue = " + byValue);
    }
}

RemoveComparison.main(null);
````
